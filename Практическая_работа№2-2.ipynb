Задание 2.5.
Высокая корреляция не доказывает наличие причинно-следственной связи, т.к. кореляция может происходить из-за третьего и более внешних факторов. Например: Продажа мороженого на самом деле не приводит к увеличению количества несчастных случаев на воде. Если учесть третий фактор- жаркая погода, то получается, что именно он являетс решающим.

```python
#ЗАДАНИЕ 3.1.
import numpy as np
x = np.array([28, 31, 29, 45, 32, 30])
SrZnach=np.mean(x)
print(SrZnach)

np.random.seed(42)
B = 5000
means = []

for i in range(B):
    sample = np.random.choice(x, size=len(x), replace=True)
    means.append(sample.mean())

means = np.array(means)

low = np.quantile(means, 0.025)
high = np.quantile(means, 0.975)

print("95% доверительный интервал:", low, "-", high)
```

Задание 3.2

H0: μB = μA. Новый алгоритм не влияет на среднее время доставки, то есть время доставки не уменьшилось.

Альтернативная гипотеза

H1: μB < μA. С новым алгоритмом среднее время доставки меньше, чем в контрольной группе.

Ошибки I и II рода

Ошибка I рода — это когда мы отвергаем H0, хотя она на самом деле верна. В нашем эксперименте это значит, что мы решили, что новый алгоритм уменьшает время доставки. На самом деле эффекта нет, а разница получилась случайно.

Ошибка II рода — это когда мы не отвергаем H0, хотя на самом деле верна H1. То есть новый алгоритм действительно уменьшает время доставки, но тест этого не показал, и мы решили, что эффекта нет.

Ложная тревога и пропущенный эффект

Ложная тревога — это ошибка I рода: мы нашли эффект, которого нет.

Пропущенный эффект — это ошибка II рода: мы не заметили эффект, который есть.

Ситуация для анализа

К прямым напрасным расходам может привести ошибка I рода. Если мы ошибочно решим, что алгоритм работает, компания внедрит его и будет платить 5 млн рублей в год. При этом время доставки на самом деле не уменьшится, и деньги будут потрачены зря.

При ошибке II рода компания не тратит деньги на алгоритм, а просто упускает возможную выгоду от его внедрения. Это тоже плохо, но это не прямые расходы, а упущенная выгода.

```python
#Задание 3.3
p_value = 0.001
alpha = 0.05

if p_value < alpha:
    print("Результат статистически значим")
else:
    print("Результат не значим")
```
